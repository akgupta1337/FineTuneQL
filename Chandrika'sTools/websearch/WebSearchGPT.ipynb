```python
from fpdf import FPDF

# Replace curly apostrophes with straight ones to avoid encoding issues
def sanitize_text(text):
    return text.replace("’", "'").replace("“", '"').replace("”", '"')

# Create a PDF object
pdf = FPDF()
pdf.add_page()
pdf.set_auto_page_break(auto=True, margin=15)

# Title
pdf.set_font("Arial", 'B', 14)
pdf.cell(0, 10, sanitize_text("Recognition of Prior Learning (RPL-NonCore)"), ln=True, align='C')
pdf.ln(10)

# Course Code
pdf.set_font("Arial", 'B', 12)
pdf.cell(0, 10, sanitize_text("Course Code: ANALYTICAL SKILLS-II"), ln=True)
pdf.ln(5)


# Save the PDF
output_path = "RPL_NonCore_Analytical_Skills_II.pdf"
pdf.output(output_path)

output_path
```
