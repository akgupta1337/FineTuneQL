```python
def findPairs(text):
    pairs = {}
    i=0
    while(i < len(text)-1):
        pair = (text[i] ,text[i+1])
        pairs[pair] = pairs.get(pair, 0) + 1
        i+=1
    return pairs
```

```python
def replace(max_pair, ord_text, new_token):
    j=0
    i=0
    temp = []
    while(i < len(ord_text)):
        if(i < len(ord_text)-1 and ord_text[i] == max_pair[0] and ord_text[i+1] == max_pair[1]):
            temp.append(new_token)
            i += 2
            continue
            
        temp.append(ord_text[i])
        i += 1
    return temp
```

```python
with open("poem.txt", "r", encoding="utf-8") as f:
    sentences = []
    while(True):
        s = f.readline()
        if(s == "<endOfFile>"):
            break
        if(s.isspace()):
            continue
        sentences.append(s.strip())
    
total = (len(((sentences))))
print(total)
```

```python
vocab = {}
for i in text2:
    vocab[ord(i)] = i

k=0

for text in sentences:
    ord_text = []
    for i in text:
        ord_text.append(ord(i))
    count = 0
    total_tokens = 10
    while(True):
        # Find Max Pair
        pairs = findPairs(ord_text)
        max_pair = max(pairs, key = lambda x: pairs[x])
        # print(ord_text)
        # print(pairs)
        # print("Max Pair: ", max_pair)
        # print()
        if(pairs[max_pair] == 1):
            break
        # Mint New Token
        new_token = total+k
        count += 1
        vocab[new_token] = vocab[max_pair[0]] + vocab[max_pair[1]] 
        k += 1
        
        # Replace
        ord_text = replace(max_pair, ord_text, new_token)
    
stoi = {s : i for i, s in vocab.items()}
```

```python
def encode(text):
    text = text.strip()
    ord_text = []
    for i in text:
        ord_text.append(ord(i))
        
    while(True):
        # Find Max Pair
        pairs = findPairs(ord_text)
        max_pair = max(pairs, key = lambda x: pairs[x])
        # print(ord_text)
        # print(pairs)
        # print("Max Pair: ", max_pair)
        if(pairs[max_pair] == 1):
            break
        max_text = vocab[max_pair[0]] + vocab[max_pair[1]]
        new_token = stoi[max_text]
        # Replace
        ord_text = replace(max_pair, ord_text, new_token)
    return ord_text
```
